```python
# Inside scripts/clean.py or your notebook
def clean_customers_df(df):
    """Cleans an in-memory customer DataFrame and returns it."""
    df = df.copy()
    df = remove_duplicates(df)
    df = standardize_categories(df)
    df = label_missing(df)
    return df

# Now you can use it cleanly on in-memory data:
cleaned_customers = clean_customers_df(customers_raw)
```
